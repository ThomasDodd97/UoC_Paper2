```python
from rdkit.Chem import AllChem as AllChem
from rdkit.Chem import Draw
from PIL import Image
from io import BytesIO
```

```python
rxn = AllChem.ReactionFromSmarts('C=C(CC(=O)O)C(=O)O.CO>CC1=CC=C(C=C1)S(=O)(=O)O>COC(=O)CC(=C)C(=O)OC .O',useSmiles=True)
d2d = Draw.MolDraw2DCairo(1000,400)
dopts = d2d.drawOptions()
dopts.useBWAtomPalette()
d2d.DrawReaction(rxn)
d2d.FinishDrawing()
bio = BytesIO(d2d.GetDrawingText())
Image.open(bio)
```

```python
rxn = AllChem.ReactionFromSmarts('CC(CO)O.C=C(CC(=O)O)C(=O)O>CC1=CC=CC=C1C.CC1=CC(=CC=C1)C.CC1=CC=C(C=C1)C>CC(COC(=O)CC(=C)C(=O)O-*)O-*.O',useSmiles=True)
d2d = Draw.MolDraw2DCairo(1300,400)
dopts = d2d.drawOptions()
dopts.useBWAtomPalette()
d2d.DrawReaction(rxn)
d2d.FinishDrawing()
bio = BytesIO(d2d.GetDrawingText())
Image.open(bio)
```
