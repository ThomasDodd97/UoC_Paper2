```python
with schemdraw.Drawing() as d:

    BioBased = '#588157'
    Styrene = '#a71e34'

    ObjectiveNaive_c = "#577590"
    UniObjective_c = "#f8961e"
    DiObjective_c = "#f3722c"

    d.config(fontsize=10,unit=1)

    OptWidth = 1.2
    OptHeight = 1
    DataWidth = 1.1
    DataHeight = 1.1

    Opt_Sobol = flow.Process(w=1,h=1).label(r'$\underset{[1-8]}{\overset{sobol}{\emptyset}}$').color(ObjectiveNaive_c)

    flow.Arrow(arrowwidth=0.01).right(d.unit*0.5).at(Opt_Sobol.E).linewidth(0.01).color('white') # hacky gap
    DatasetOne = flow.Start(w=DataWidth,h=DataHeight).label(r'$\underset{[1-8]}{\mathcal{D}}$').anchor('W')

    flow.Wire('-|', arrow='->').at(Opt_Sobol.E).to(DatasetOne.W).linestyle("-")

    # -=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=

    flow.Arrow(arrowwidth=0.01).right(d.unit*1).at(DatasetOne.E).linewidth(0.01).color('white') # hacky gap
    flow.Arrow(arrowwidth=0.01).up(d.unit*1.2).linewidth(0.01).color('white') # hacky gap
    Opt_One_MIPT = flow.Process(w=OptWidth,h=OptHeight).label(r'$\underset{[9-24]}{\overset{MIPT}{\emptyset}}$').anchor('W').color(ObjectiveNaive_c)

    flow.Arrow(arrowwidth=0.01).right(d.unit*1).at(DatasetOne.E).linewidth(0.01).color('white') # hacky gap
    Opt_One_YBP = flow.Process(w=OptWidth,h=OptHeight).label(r'$\underset{[25-40]}{\overset{qLNEI}{\sigma_{\text{ybp}}}}$').anchor('W').color(UniObjective_c)

    flow.Arrow(arrowwidth=0.01).right(d.unit*1).at(DatasetOne.E).linewidth(0.01).color('white') # hacky gap
    flow.Arrow(arrowwidth=0.01).down(d.unit*1.2).linewidth(0.01).color('white') # hacky gap
    Opt_One_E = flow.Process(w=OptWidth,h=OptHeight).label(r'$\underset{[41-56]}{\overset{qLNEI}{E}}$').anchor('W').color(UniObjective_c)

    flow.Wire('c',k=0.5,arrow='->').at(DatasetOne.E).to(Opt_One_MIPT.W).linestyle("-")
    flow.Wire('c',k=0.5,arrow='->').at(DatasetOne.E).to(Opt_One_YBP.W).linestyle("-")
    flow.Wire('c',k=0.5,arrow='->').at(DatasetOne.E).to(Opt_One_E.W).linestyle("-")

    # -=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=

    flow.Arrow(arrowwidth=0.01).right(d.unit*1).at(Opt_One_YBP.E).linewidth(0.01).color('white') # hacky gap
    DatasetTwo = flow.Start(w=DataWidth,h=DataHeight).label(r'$\underset{[1-56]}{\mathcal{D}}$').anchor('W')

    flow.Wire('c',k=0.5,arrow='->').at(Opt_One_MIPT.E).to(DatasetTwo.W).linestyle("-")
    flow.Wire('c',k=0.5,arrow='->').at(Opt_One_YBP.E).to(DatasetTwo.W).linestyle("-")
    flow.Wire('c',k=0.5,arrow='->').at(Opt_One_E.E).to(DatasetTwo.W).linestyle("-")
    flow.Wire('n',k=1.5,arrow="->").at(DatasetOne.N).to(DatasetTwo.N)

    # -=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=

    flow.Arrow(arrowwidth=0.01).right(d.unit*0.5).at(DatasetTwo.E).linewidth(0.01).color('white') # hacky gap
    Opt_Two_YBPvE = flow.Process(w=OptWidth*1.3,h=OptHeight).label(r'$\underset{[57-72]}{\overset{qLNEHVI}{\sigma_{\text{ybp}}\ \text{vs}\ E}}$').anchor('W').color(DiObjective_c)

    flow.Wire('c',k=0.5,arrow='->').at(DatasetTwo.E).to(Opt_Two_YBPvE.W).linestyle("-")

    # -=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=

    flow.Arrow(arrowwidth=0.01).right(d.unit*0.5).at(Opt_Two_YBPvE.E).linewidth(0.01).color('white') # hacky gap
    DatasetThree = flow.Start(w=DataWidth,h=DataHeight).label(r'$\underset{[1-72]}{\mathcal{D}}$').anchor('W')
    flow.Wire('c',k=0.5,arrow='->').at(Opt_Two_YBPvE.E).to(DatasetThree.W).linestyle("-")
    flow.Wire('n',k=-0.65,arrow="->").at(DatasetTwo.S).to(DatasetThree.S)

    # -=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=

    flow.Arrow(arrowwidth=0.01).down(d.unit*2).at(Opt_Sobol.S).linewidth(0.01).color('white') # hacky gap
    SobolStyrene = flow.Process(w=1,h=1).label(r'$\underset{[73-80]}{\overset{sobol}{\emptyset}}$').anchor('N').color(ObjectiveNaive_c)

    flow.Arrow(arrowwidth=0.01).right(d.unit*0.5).at(SobolStyrene.E).linewidth(0.01).color('white') # hacky gap
    DatasetFour = flow.Start(w=DataWidth,h=DataHeight).label(r'$\underset{[73-80]}{\mathcal{D}}$').anchor('W')

    flow.Arrow(arrowwidth=0.01).right(d.unit*1).at(DatasetFour.E).linewidth(0.01).color('white') # hacky gap
    Opt_Four_E = flow.Process(w=OptWidth,h=OptHeight).label(r'$\underset{[81-96]}{\overset{qLNEI}{E}}$').anchor('W').color(UniObjective_c)

    flow.Arrow(arrowwidth=0.01).right(d.unit*1).at(Opt_Four_E.E).linewidth(0.01).color('white') # hacky gap
    DatasetFive = flow.Start(w=DataWidth,h=DataHeight).label(r'$\underset{[73-96]}{\mathcal{D}}$').anchor('W')

    flow.Wire('c',k=0.5,arrow='->').at(SobolStyrene.E).to(DatasetFour.W).linestyle("-")
    flow.Wire('c',k=0.5,arrow='->').at(DatasetFour.E).to(Opt_Four_E.W).linestyle("-")
    flow.Wire('c',k=0.5,arrow='->').at(Opt_Four_E.E).to(DatasetFive.W).linestyle("-")
    flow.Wire('n',k=-0.4,arrow="->").at(DatasetFour.S).to(DatasetFive.S)

    # -=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=

    flow.Arrow(arrowwidth=0.01).left(d.unit*0.4).at(Opt_Sobol.W).linewidth(0.01).color('white') # hacky gap
    flow.Arrow(arrowwidth=0.01).up(d.unit*0.15).linewidth(0.01).color('white') # hacky gap
    flow.Arrow(arrowwidth=0.01).right(d.unit*0.25).linewidth(0.01).color('white') # hacky gap
    flow.Box(w=10.9,h=4.1).linestyle('--').color(BioBased).label("Dimethyl Itaconate",rotate=90,ofst=(-5.75,0))

    flow.Arrow(arrowwidth=0.01).left(d.unit*0.4).at(SobolStyrene.W).linewidth(0.01).color('white') # hacky gap
    flow.Arrow(arrowwidth=0.01).down(d.unit*0.2).linewidth(0.01).color('white') # hacky gap
    flow.Arrow(arrowwidth=0.01).right(d.unit*0.25).linewidth(0.01).color('white') # hacky gap
    flow.Box(w=7.2,h=2).linestyle('--').color(Styrene).label("Styrene",rotate=90,ofst=(-3.9,0))

    # -=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=

    flow.Arrow(arrowwidth=0.01).down(d.unit*0.85).at(DatasetFour.S).linewidth(0.01).color('white') # hacky gap
    flow.Arrow(arrowwidth=0.01).right(d.unit*1.2).linewidth(0.01).color('white') # hacky gap
    Legend1 = flow.Start(w=2.4,h=1).anchor('N').label(r'$\underset{[\text{Samples Contained}]}{Dataset, \mathcal{D}}$')

    flow.Arrow(arrowwidth=0.01).down(d.unit*0.01).at(Legend1.E).linewidth(0.01).color('white') # hacky gap
    flow.Arrow(arrowwidth=0.01).right(d.unit*1.5).linewidth(0.01).color('white') # hacky gap
    Legend2 = flow.Process(w=2.4,h=1).anchor('W').label(r'$\underset{[\text{Samples Produced}]}{\overset{Sampler}{Objective(s)}}$')

    flow.Arrow(arrowwidth=0.01).up(d.unit*0.75).at(DatasetFive.E).linewidth(0.01).color('white') # hacky gap
    flow.Arrow(arrowwidth=0.01).right(d.unit*2.1).linewidth(0.01).color('white') # hacky gap
    Legend3 = flow.Process(w=3.2,h=0.5,color=ObjectiveNaive_c).label('Objective Naive').anchor('N')

    flow.Arrow(arrowwidth=0.01).down(d.unit*0.2).at(Legend3.S).linewidth(0.01).color('white') # hacky gap
    Legend4 = flow.Process(w=3.2,h=0.5,color=UniObjective_c).label('Uni-Objective').anchor('N')

    flow.Arrow(arrowwidth=0.01).down(d.unit*0.2).at(Legend4.S).linewidth(0.01).color('white') # hacky gap
    Legend5 = flow.Process(w=3.2,h=0.5,color=DiObjective_c).label('Di-Objective').anchor('N')

    d.save('OptimiProcdr.png',transparent=False,dpi=1000)
```
