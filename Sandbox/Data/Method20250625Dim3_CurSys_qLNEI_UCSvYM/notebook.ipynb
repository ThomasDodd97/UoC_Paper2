```python
# Having identified sample 22 as a particularly succesful low calcium sulphate sample we will look at its
# formulation and see what lies behind.

x1_22 = backup["x1"][22]
x2_22 = backup["x2"][22]
x3_22 = backup["x3"][22]

i1dbp_22 = backup["a5"][22]
i1cs_22 = backup["a4"][22]
cs_22 = backup["a9"][22]
upr1_22 = backup["a8"][22]
i2_22 = backup["a10"][22]

dbp_22 = backup["b1"][22]
dmi_22 = backup["b2"][22]
cs_22 = backup["b3"][22]
up1_22 = backup["b4"][22]

pct_dbp_22 = (dbp_22 / (dmi_22+cs_22+dbp_22+up1_22))*100
pct_dmi_22 = (dmi_22 / (dmi_22+cs_22+dbp_22+up1_22))*100
pct_cs_22 = (cs_22 / (dmi_22+cs_22+dbp_22+up1_22))*100
pct_up1_22 = (up1_22 / (dmi_22+cs_22+dbp_22+up1_22))*100

print("Sample 22")
print()
print(f"x1 = {round(x1_22,2)}")
print(f"x2 = {round(x2_22,2)}")
print(f"x3 = {round(x3_22,2)}")
print()
print(f"DMI = {round(pct_dmi_22)}%")
print(f"CS = {round(pct_cs_22)}%")
print(f"DBP = {round(pct_dbp_22)}%")
print(f"UP1 = {round(pct_up1_22)}%")
print()
print(f"UPR1 = {round(pct_dmi_22+pct_up1_22)}%")
print(f"Filler = {round(pct_cs_22)}%")
print(f"Initiator = {round(pct_dbp_22)}%")
print()
print(f"Actual Mass of DBP Powder = {round(i1dbp_22+i1cs_22,2)} g")
print(f"Actual Mass of CS Powder = {round(cs_22,2)} g")
print(f"Actual Mass of UPR1 = {round(upr1_22,2)} g")
print()
print(f"Actual Percent DBP Powder = {round(((i1dbp_22+i1cs_22)/(i1dbp_22+i1cs_22+cs_22+upr1_22))*100,2)} %")
print(f"Actual Percent CS Powder = {round((cs_22/(i1dbp_22+i1cs_22+cs_22+upr1_22))*100,2)} %")
print(f"Actual Percent UPR1 = {round((upr1_22/(i1dbp_22+i1cs_22+cs_22+upr1_22))*100,2)} %")
```
